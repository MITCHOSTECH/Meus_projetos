```python
quadrados = []
```

```python
type(quadrados)
```

```python
for x in range(10):
    quadrados.append(x ** 2)
print(f"Utilizar ciclo for tradicional: {quadrados}")
```

```python
quadrados_2 = [x ** 2 for x in range(10)]
print(f"Quadrados utilizando uma lista comprehensions {quadrados_2} ")
```

```python
###COMPREENSÕES DE CONDIÇÃO -> UTILIZAR IF e ELSE
```

```python
# INCLUIR APENAS MULTIPLOS DE 3 ENTRE 0 e 30
# Multiplos de três quer dizer números divisíveis a 3 com resultado 0
multiplos_3 = [x  for x in range(30) if x % 3 == 0]
print(f"Multiplos de três: {multiplos_3}")
```

```python
### PAR OU IMPAR UTILIZANDO CONDIÇÃO ELSE
par_impar = ['Par' if x % 2 == 0 else 'Impar' for x in range(11)]
print(f"Par ou Impar: {par_impar}")
```
